# Classes

Classes provide a means of bundling data and functionality together. 

Creating a new class creates a new type of object, allowing new instances of that type to be made. 

Each class instance can have attributes attached to it for maintaining its state. Class instances can also have methods (defined by its class) for modifying its state.

 Python classes provide all the standard features of Object Oriented Programming: the class inheritance mechanism allows multiple base classes, a derived class can override any methods of its base class or classes, and a method can call the method of a base class with the same name. Objects can contain arbitrary amounts and kinds of data. As is true for modules, classes partake of the dynamic nature of Python: they are created at runtime, and can be modified further after creation.

 OR

 Object-Oriented Programming (OOP) is a programming paradigm that organizes code using objects.

# What is a Class?

A class is a blueprint or template for creating objects.

It defines the attributes and methods that objects created from it will have.

# What is an Object?

An object is an instance of a class.

If a class is a blueprint, then an object is the actual thing created from that blueprint.

````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
````

# The `__init__()` Method (Constructor)

The `__init__()` method is a special method that runs automatically whenever an object is created.

It is used to initialize (give initial values to) the object's attributes.

# Class Objects

Class objects support two kinds of operations: `attribute references `and `instantiation`.

Attribute references use the standard syntax used for all attribute references in Python: `obj.name`.

Valid attribute names are all the names that were in the class’s namespace when the class object was created. 


````python
class MyClass:
    """A simple example class"""
    i = 12345

    def f(self):
        return 'hello world'
````

````python
MyClass.i
````

````python
MyClass.f
````

then `MyClass.i` and `MyClass.f` are valid attribute references, returning an integer and a function object, respectively. 

Class instantiation uses function notation. Just pretend that the class object is a parameterless function that returns a new instance of the class.

````python
x = MyClass()
````

#  Instance Objects¶

When we create an object from a class, that object is called an instance object.

Think of a class as a blueprint and an object (instance) as the real thing created from that blueprint.

````python
class Student:
    pass

student1 = Student()
student2 = Student()
````

Here:

1. Student is a class.
2. student1 is an instance (object).
3. student2 is another instance.

Each object has its own data.

# What Can We Do With an Instance Object?

The Python documentation says:

The only operations understood by instance objects are attribute references.

This means we mainly interact with objects by using attributes.

Attributes are things that belong to an object.

There are two types of attributes:

1. Data Attributes
2. Methods

# Data Attributes (Instance Variables)

A data attribute is simply a variable that belongs to an object.

Example:

````python
class Student:
    pass

student = Student()

student.name = "Collins"
student.age = 23

print(student.name)
print(student.age)
````

Here,

1. student.name
2. student.age

are instance variables (data attributes).

Each object stores its own values.

# Real-Life Example

Imagine a class called Car.

Every car has:

1. Brand
2. Color
3. Speed

Example:

````python
class Car:
    pass

car1 = Car()
car2 = Car()

car1.brand = "Toyota"
car1.color = "Blue"

car2.brand = "Honda"
car2.color = "Red"

print(car1.brand)
print(car2.brand)
````

# Methods

A method is a function that belongs to a class and is called using an object.

````python
class Student:

    def greet(self):
        print("Hello Student")


student = Student()

# calling a method.
student.greet()
````

So, an object can have both:

Data attributes (variables like name, age, counter)

Methods (functions like greet())

# What is `__init__()`?

`__init__()` is a special method (also called a constructor or initializer) that Python automatically calls when an object is created.

It is used to initialize an object's attributes.

````python
class Student:

    def __init__(self, name, course):
        self.name = name
        self.course = course


student1 = Student("Collins", "Computer Science")
student2 = Student("Mary", "Information Technology")

print(student1.name)
print(student2.name)
````
